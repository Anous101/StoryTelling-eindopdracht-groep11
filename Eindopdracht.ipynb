<h3 style="font-weight:bold">A) Algemene trend wereldwijd</h3>

<p style="font-weight:bold">.Type:</p>Area chart (of lijndiagram)

<p style="font-weight:bold">.De inhoud :</p> Totale wereldwijde bosoppervlakte per jaar (som van alle landen)

<p style="font-weight:bold">.Waarom :</p> Een area chart benadrukt zowel de trend als de omvang van het bosoppervlak. Het gevulde vlak versterkt het gevoel van verlies.

<p style="font-weight:bold">.Clutter verwijderen:</p> Alleen wereldwijde totalen getoond, geen landen of groepen.

<p style="font-weight:bold">.preattentive attributes:</p>
Kleur: Groen want het heeft directe associatie met bos.
Vorm: Vloeiende lijn en gevuld vlak trekken aandacht naar de daling


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("forest_area_cleaned.csv")

world_trend = df.groupby("year")["forest_area"].sum().reset_index()

plt.figure(figsize=(10,6))
plt.fill_between(world_trend["year"], world_trend["forest_area"], alpha=0.4, color="darkgreen")
plt.plot(world_trend["year"], world_trend["forest_area"], color="green")
plt.title("Wereldwijde afname van bosoppervlakte (1990–2020)")
plt.ylabel("Totaal bosoppervlakte (km²)")
plt.xlabel("Jaar")
plt.show()
```

Er is iets fout met de dataset we zien dat er een hele hoge stijging van 1990 tot 1992 en dat loopt het goed vanaf 1992 met een daling tot 2020


<h3 style="font-weight:bold">B) Trend per inkomensniveau 1990-2020 :</h3>

<p style="font-weight:bold">.Type:</p> Line chart

<p style="font-weight:bold">.Inhoud:</p> Bosoppervlakte per jaar per inkomensgroep (low income, lower middle, middle, upper middle, high income)

<p style="font-weight:bold">.Waarom:</p> Lijncharts zijn ideaal om trends over tijd te vergelijken tussen meerdere categorieen.

<p style="font-weight:bold">.clutter</p> verwijderen: Alleen vijf inkomensgroepen getoond, geen landen.

<p style="font-weight:bold">Preattentive attributes:</p>

Kleur: verschillende kleuren per inkomensgroep.

Vorm: Alle lijnen lopen horizontaal over dezelfde x-as (directe vergelijking).

Lengte/richting: Daling of stijging is meteen zichtbaar.


```python
income_groups = [
    "Low income",
    "Lower middle income",
    "Middle income",
    "Upper middle income",
    "High income"
]

income_trend = df[df["area_name"].isin(income_groups)]

plt.figure(figsize=(10,6))
sns.lineplot(data=income_trend, x="year", y="forest_area", hue="area_name")
plt.title("Bosoppervlakte per inkomensniveau (1990–2020)")
plt.ylabel("Bosoppervlakte (km²)")
plt.xlabel("Jaar")
plt.legend(title="Inkomensniveau")
plt.show()

```

<h3 style="font-weight:bold">C) Trend 2018–2020 in 3 landen en de wereld</h3>

<p style="font-weight:bold">.Type: BarChart</p>

<p style="font-weight:bold">.Inhoud:</p> Bosoppervlakte per jaar voor drie landen bijv Brazilie, Indonesie, Rusland en de wereld
 
<p style="font-weight:bold">.Waarom:</p> Barchart maakt korte termijn patterns zichtbaar en laat directe vergelijking per jaar toe

<p style="font-weight:bold">.Clutter verwijderen:</p> Slechts 3 landen en "World" zijn getoond.

<p style="font-weight:bold">.Preattentive attributes:</p>

Kleur: Elke landgroep heeft een eigen kleur.

Lengte: Hoogte van de balken trekt direct aandacht naar verschillen.

Groepering: Balken per jaar naast elkaar maken vergelijking eenvoudig.


```python
landen = ["Brazil", "Indonesia", "Russian Federation", "World"]
subset = df[df["area_name"].isin(landen) & df["year"].isin([2018, 2019, 2020])]

plt.figure(figsize=(10,6))
sns.barplot(data=subset, x="year", y="forest_area", hue="area_name")
plt.title("Bosoppervlakte per land en wereld 2018–2020")
plt.ylabel("Bosoppervlakte (km²)")
plt.xlabel("Jaar")
plt.legend(title="Landen")
plt.show()
```

<h3>D) Bosoppervlakte in 2020 per inkomensniveau en de wereld</h3>

<p style="font-weight:bold">.Type:</p> Horizontale barchart

<p style="font-weight:bold">.Inhoud:</p> Bosoppervlakte per inkomensniveau in 2020. Wereldwaarde is apart berekend en vermeld.
<p style="font-weight:bold">.Waarom:</p> Zo kunnen VN-leiders de verschillen tussen inkomensgroepen zien zonder dat de enorme wereldwaarde de grafiek domineert.

<p style="font-weight:bold">.Clutters</p> verwijderen : Alle clutters zijn verwijdert, alleen inkomensgroepen getoond, geen landen.

<p style="font-weight:bold">.Preattentive attributes:</p>

<p style="font-weight:bold">Kleur:</p> Elke landgroep heeft een eigen kleur.
<br>
<p style="font-weight:bold">Lengte:</p> Hoogte van de balken trekt direct aandacht naar verschillen.
<br>
<p style="font-weight:bold">Groepen:</p> Balken per jaar naast elkaar maken vergelijking eenvoudig.
<br>
<br>
<p style="font-weight:bold">Totale bosoppervlakte wereldwijd in 2020:</p> 423268566.2 km²


```python
latest_year = 2020
world_2020 = df[df["year"] == latest_year]["forest_area"].sum()
print("Totale bosoppervlakte wereldwijd in 2020:", world_2020)

income_groups = [
    "Low income",
    "Lower middle income",
    "Middle income",
    "Upper middle income",
    "High income"
]

income_2020 = df[(df["year"] == latest_year) & (df["area_name"].isin(income_groups))]

plt.figure(figsize=(8,6))
sns.barplot(data=income_2020, x="forest_area", y="area_name", palette="Greens_r")
plt.title("Bosoppervlakte per inkomensniveau 2020")
plt.xlabel("Bosoppervlakte (km²)")
plt.ylabel("Inkomensniveau")
plt.show()
```

<h3>E) Gemiddelde wereldtemperatuur 1995-2019</h3>
<br>
<p style="font-weight:bold">.Type :</p> Line Chart
<br>
<br>
<p style="font-weight:bold">.Inhoud:</p> Gemiddeld temperatuur wereldwijd per jaar (alle landen)
<br>
<br>
<p style="font-weight:bold">.Waarom:</p> Laat de lange termijn trend zien in de wereldwijde temperatuur,dit maakt duidelijk of er sprake is van opwarming.
<br>
<br>
<p style="font-weight:bold">.Clutter</p> verwijderen : Alleen de wereldgemiddelde per jaar getoond, geen landen of steden.
<br>
<br>
<p style="font-weight:bold">.Preattentive attributes:</p>   
<br>
<br>
<p style="font-weight:bold">Kleur:</p> Rood die heeftdirecte associatie met warmte en klimaatverandering.
<br>
<p style="font-weight:bold">Positie:</p> De stijgende lijn op de y-as trekt direct aandacht.
<br>
<p style="font-weight:bold">Orientatie:</p> Tijd loopt horizontaal, waardoor de trend intuitief te volgen is.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("city_temperature_preprocessed.csv")

world = df[(df["Year"] >= 1995) & (df["Year"] <= 2019)]

world_trend = world.groupby("Year")["AvgTemperature"].mean().reset_index()

plt.figure(figsize=(10,6))
sns.lineplot(data=world_trend, x="Year", y="AvgTemperature", color="red")
plt.title("Gemiddelde wereldtemperatuur 1995–2019")
plt.ylabel("Gemiddelde temperatuur (°C)")
plt.xlabel("Jaar")
plt.show()
```

<h3>F) Gemiddled temperatuur per maand in Nederland</h3>
<br>
<br>
<p style="font-weight:bold">.Type:</p> Boxplot / per maand
<br><br>
<p style="font-weight:bold">.inhoud:</p> Voor elke maand een boxplot die de dagelijkse gemiddlede tempratuur in Nederland laat zien.
<br><br>
<p style="font-weight:bold">.Waarom:</p> Boxplots toenen spreiding en vriatie en dat maakt de seizoenenpatronen en extreme waarden makkelijker te zien.
<br><br>
<p style="font-weight:bold">.Clutter</p> verwijderen: Alleen Nederland gefilterd
<br><br>
<p style="font-weight:bold">.Preattentive attributes:</p>   
<br><br>
<p style="font-weight:bold">Kleur:</p>  (blauw/rood) die direct associatie heeft met koud en warm en versterkt de seizoenscontrast.
<br>
<p style="font-weight:bold">Lengte:</p> Hoogte van de boxen maakt verschillen tussen maanden direct zichtbaar.
<br>
<p style="font-weight:bold">Positie:</p> Maanden op de x-as in volgorde  (intuitieve vergelijking).



```python
Netherlands = df[df["Country"] == "The Netherlands"]

plt.figure(figsize=(12,6))
sns.boxplot(data = Netherlands, x="Month", y="AvgTemperature", palette="coolwarm")
plt.title("Gemiddelde temperatuur per maand in Nederland")
plt.ylabel("Gemiddelde temperatuur (°C)")
plt.xlabel("Maand")
plt.show()
```
